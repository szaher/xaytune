## Alignment

Align models with human preferences using `align()`. Supports multiple alignment methods:

- **DPO** (Direct Preference Optimization)
- **GRPO** (Group Relative Policy Optimization)
- **PPO** (Proximal Policy Optimization)
- **RLAIF** (Reinforcement Learning from AI Feedback)

Alignment typically follows supervised fine-tuning (SFT) and uses preference datasets.

## Pre-Training

Use `pretrain()` for continued pre-training on raw text corpora. This is ideal for domain adaptation or training on proprietary data.

The function handles tokenization, data loading, and distributed training automatically.

```python
import trainlib

# LoRA fine-tuning in one line
state = trainlib.finetune(
    model="meta-llama/Llama-3.1-8B",
    dataset="data/train.jsonl",
    method="lora",
    format="alpaca",
    num_epochs=3,
)

# Inspect results
print(f"Final loss: {state.metrics.get('loss', 'N/A')}")
print(f"Total steps: {state.global_step}")
```

## Fine-Tuning

`finetune()` is the simplest way to fine-tune an LLM on instruction or chat data. It supports multiple parameter-efficient methods (LoRA, QLoRA) and full fine-tuning.

The function returns a `TrainingState` object containing metrics, the trained model state, and training history.
